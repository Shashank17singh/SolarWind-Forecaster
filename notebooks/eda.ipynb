# Phase 1: Exploratory Data Analysis (EDA)


This module performs statistical analysis and data validation on high-resolution (1-minute) solar wind telemetry (OMNI dataset). The primary objective is to evaluate the distributional properties of geomagnetic storm indicators, specifically the SYM-H index, to inform downstream feature engineering and model selection.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("viridis")
```

## 1.1 Data Ingestion and Telemetry Parsing
The OMNI dataset provides multivariate time-series features, including Interplanetary Magnetic Field (IMF) magnitude, solar wind velocity components, and proton density measurements.

```python
try:
    df = pd.read_csv('../data/processed/omni.csv', parse_dates=['timestamp'])
    print(f"Loaded {len(df)} records.")
    display(df.head())
except FileNotFoundError:
    print("Dataset not found. Please run src/data_ingestion.py first to download and parse the raw data.")
```

## 1.2 Geomagnetic Storm Proxy (SYM-H) Validation
The SYM-H index serves as the primary target variable. Thresholds of -50 nT and -100 nT correspond to moderate and intense geomagnetic storm classifications, respectively.

```python
if 'df' in locals():
    plt.figure(figsize=(12, 6))
    sns.histplot(df['sym_h'], bins=100, kde=True)
    plt.axvline(-50, color='r', linestyle='--', label='Moderate Storm (-50 nT)')
    plt.axvline(-100, color='darkred', linestyle='--', label='Intense Storm (-100 nT)')
    plt.title('Distribution of SYM-H Index')
    plt.xlabel('SYM-H (nT)')
    plt.ylabel('Frequency')
    plt.legend()
    plt.show()
```
