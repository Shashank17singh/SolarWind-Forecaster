## 3.2 Model Interpretability and Feature Importance
Post-training analysis leverages tree-based feature importance metrics to identify the most critical drivers of geomagnetic anomalies.

```python
import joblib

try:
    model = joblib.load("../models/symh_model.joblib")
    lgb.plot_importance(
        model, max_num_features=15, figsize=(10, 8), title="LightGBM Feature Importance"
    )
    plt.show()
except FileNotFoundError:
    print("Model not trained yet.")
```
